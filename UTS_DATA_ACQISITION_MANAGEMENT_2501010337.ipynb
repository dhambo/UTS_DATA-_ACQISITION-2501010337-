## Soal 1: Akuisisi 1 (Unduh Dataset), Identifikasi Struktur, & Estimasi Storage

### 1. Sumber Data & Metode Akuisisi
Sumber data pertama diperoleh melalui metode **unduhan berkas terstruktur (`CSV`)** dari halaman resmi Kaggle: https://www.kaggle.com/datasets/sagitasantia/wisata-indonesia (berkas `wisata_nusantara.csv`). Berkas mentah disimpan utuh di folder `data raw/` tanpa mengubah isi maupun struktur aslinya.

### 2. Tujuan & Langkah Pemrosesan
* Menghitung **hash MD5** dan ukuran berkas mentah (`Bytes` dan `KB`) sebagai bukti integritas data (*provenance*).
* Membaca dataset menggunakan parameter `sep=","` dan `encoding="utf-8"`, lalu mengidentifikasi dimensi baris, kolom, dan tipe data.
* Menghitung **estimasi kebutuhan penyimpanan (*storage*) 1 tahun** dengan rumus `(ukuran berkas × 12 bulan) + cadangan 30%`.
* Membersihkan spasi berlebih pada kolom wilayah (`nama_wisata`, `provinsi`, `kota_kabupaten`) dan menyimpannya ke `process/wisata_nusantara_bersih.csv`.
*

```python
import os
import shutil
import hashlib
import pandas as pd

os.makedirs("data raw", exist_ok=True)
os.makedirs("process", exist_ok=True)

raw_csv_path = "data raw/wisata_nusantara.csv"
if os.path.exists("wisata_nusantara.csv"):
    shutil.move("wisata_nusantara.csv", raw_csv_path)

if not os.path.exists(raw_csv_path):
    raise FileNotFoundError("Silakan klik ikon Folder di kiri Colab -> klik ikon Upload -> pilih 'wisata_nusantara.csv' dulu!")

with open(raw_csv_path, "rb") as f:
    md5_sumber1 = hashlib.md5(f.read()).hexdigest()

ukuran_bytes = os.path.getsize(raw_csv_path)
ukuran_kb = ukuran_bytes / 1024

print("=== PROVENANCE SUMBER 1 (CATAT KE README) ===")
print(f"Nama Berkas   : wisata_nusantara.csv")
print(f"URL Sumber    : https://www.kaggle.com/datasets/sagitasantia/wisata-indonesia")
print(f"Tanggal Akses : 2026-11-24")
print(f"Lisensi       : Other (Academic Research & Tourism Promotion - Kaggle)")
print(f"MD5 Hash      : {md5_sumber1}")
print(f"Ukuran Berkas : {ukuran_bytes:,} Bytes ({ukuran_kb:.2f} KB)\n")

df_raw = pd.read_csv(raw_csv_path, sep=",", encoding="utf-8")
print("=== IDENTIFIKASI DATASET (SOAL 1) ===")
print(f"Jenis Data    : Data Terstruktur (Tabular CSV)")
print(f"Jumlah Baris  : {df_raw.shape} baris")
print(f"Jumlah Kolom  : {df_raw.shape} kolom")
print("\nTipe Data Setiap Kolom:")
print(df_raw.dtypes)

storage_12_bln = ukuran_kb * 12
cadangan_30 = storage_12_bln * 0.30
total_storage = storage_12_bln + cadangan_30

print("\n=== ESTIMASI STORAGE 1 TAHUN ===")
print(f"12 Bulan (ukuran x 12) : {storage_12_bln:.2f} KB")
print(f"Cadangan 30%           : {cadangan_30:.2f} KB")
print(f"Total Kebutuhan 1 Thn  : {total_storage:.2f} KB ({total_storage/1024:.2f} MB)")

display(df_raw.head())
```

## Soal 3: Akuisisi Data via REST API (Open-Meteo Elevation & Weather API)

### 1. Sumber API & Nilai Informasi Data
Untuk menggantikan tarikan API sebelumnya yang kurang bernilai, akuisisi Sumber 2 menggunakan **Open-Meteo Public API** (`https://open-meteo.com`, lisensi CC BY 4.0) tanpa memerlukan API key rahasia. API ini dipanggil berdasarkan titik koordinat `latitude` dan `longitude` dari dataset wisata untuk mengambil parameter numerik yang bernilai dan bebas nilai kosong:
* **Elevasi ketinggian wilayah (`elevasi_mdpl`)**
* **Suhu udara terkini (`suhu_celcius`)**
* **Kelembapan udara (`kelembapan_persen`)**
* **Kecepatan angin (`kecepatan_angin_kmh`)**

### 2. Penerapan Etika Request Web
* Menyertakan `User-Agent` berisi email kampus resmi (`2501010337@instiki.ac.id`).
* Menggunakan `timeout=20`, jeda etis (`time.sleep(1.5)`), serta mekanisme `Retry` otomatis jika terjadi gangguan jaringan.


```python
import json
import time
import hashlib
import requests
import pandas as pd
from requests.adapters import HTTPAdapter
from urllib3.util.retry import Retry

session = requests.Session()
retries = Retry(total=3, backoff_factor=1, status_forcelist=(429, 500, 502, 503, 504))
session.mount("https://", HTTPAdapter(max_retries=retries))

headers = {
    "User-Agent": "UTS-DataAcquisition-Student/1.0 (2501010337@instiki.ac.id)"
}

df_lokasi = df_wisata_bersih.copy()
df_lokasi["latitude"] = pd.to_numeric(df_lokasi["latitude"], errors="coerce")
df_lokasi["longitude"] = pd.to_numeric(df_lokasi["longitude"], errors="coerce")
df_lokasi["kunci_wilayah"] = (
    df_lokasi["kota_kabupaten"]
    .replace("nan", pd.NA)
    .dropna()
    .str.replace(r"^(Kabupaten|Kota)\s+", "", regex=True)
    .str.strip()
    .str.title()
)

df_lokasi_unik = (
    df_lokasi.dropna(subset=["kunci_wilayah", "latitude", "longitude"])
    .drop_duplicates(subset=["kunci_wilayah"])
    .head(40)
    .reset_index(drop=True)
)

lat_str = ",".join(df_lokasi_unik["latitude"].round(4).astype(str))
lon_str = ",".join(df_lokasi_unik["longitude"].round(4).astype(str))

url_api = "https://api.open-meteo.com/v1/forecast"
params = {
    "latitude": lat_str,
    "longitude": lon_str,
    "current": "temperature_2m,relative_humidity_2m,wind_speed_10m",
    "timezone": "Asia/Makassar"
}

time.sleep(1.5)
resp_api = session.get(url_api, params=params, headers=headers, timeout=20)

if resp_api.status_code == 200:
    raw_api_data = resp_api.json()
    for i, item in enumerate(raw_api_data):
        item["kunci_wilayah"] = df_lokasi_unik.loc[i, "kunci_wilayah"]

    raw_json_path = "data raw/api_cuaca_elevasi_raw.json"
    with open(raw_json_path, "w", encoding="utf-8") as f:
        json.dump(raw_api_data, f, indent=4, ensure_ascii=False)

    with open(raw_json_path, "rb") as f:
        md5_sumber2 = hashlib.md5(f.read()).hexdigest()

    print("=== PROVENANCE SUMBER 2 (API OPEN-METEO) ===")
    print(f"Berkas Mentah : {raw_json_path}")
    print(f"URL API       : {url_api}")
    print(f"Tanggal Akses : 2026-11-27")
    print(f"Lisensi       : CC BY 4.0 (Open-Meteo Public API)")
    print(f"MD5 Hash      : {md5_sumber2}\n")

    df_norm = pd.json_normalize(raw_api_data)
    df_api_bersih = df_norm[[
        "kunci_wilayah",
        "elevation",
        "current.temperature_2m",
        "current.relative_humidity_2m",
        "current.wind_speed_10m",
        "current.time"
    ]].rename(columns={
        "elevation": "elevasi_mdpl",
        "current.temperature_2m": "suhu_celcius",
        "current.relative_humidity_2m": "kelembapan_persen",
        "current.wind_speed_10m": "kecepatan_angin_kmh",
        "current.time": "waktu_pantau"
    })

    df_api_bersih.to_csv("process/api_cuaca_elevasi_bersih.csv", index=False, encoding="utf-8")
    print(f"Jumlah Baris API Bersih: {len(df_api_bersih)} baris (0 nilai kosong)")
    display(df_api_bersih.head(10))
else:
    print(f"Gagal API. Status: {resp_api.status_code}")
```

## Soal 4: Web Scraping Etis (`RobotFileParser().can_fetch()`) & Administrasi Provinsi Indonesia

### 1. Pemeriksaan Izin Akses (`robots.txt`)
Sesuai aturan etika *web scraping*, pemeriksaan izin akses tidak dilakukan secara manual, melainkan secara terprogram menggunakan modul standar Python `urllib.robotparser.RobotFileParser()` dengan memanggil metode **`rp.can_fetch(user_agent, url_target)`**.

### 2. Perluasan Target Scraping ke Skala Nasional
Target halaman statis diperluas ke **Daftar Provinsi di Indonesia** (`https://wikipedia.org`, lisensi CC BY-SA 4.0) dan diekstrak menggunakan `BeautifulSoup` serta `pandas` sehingga informasi ibu kota, luas wilayah, dan jumlah penduduk mencakup seluruh provinsi pada dataset wisata (tidak terbatas pada wilayah Bali saja).

```python
import time
import hashlib
import requests
import pandas as pd
from bs4 import BeautifulSoup
from io import StringIO
from urllib.robotparser import RobotFileParser

user_agent = "UTS-DataAcquisition-Student/1.0 (2501010337@instiki.ac.id)"
headers = {"User-Agent": user_agent}

domain = "id.wikipedia.org"
url_robots = "https://" + domain + "/robots.txt"
url_scrape = "https://" + domain + "/wiki/Provinsi_di_Indonesia"

resp_robots = requests.get(url_robots, headers=headers, timeout=15)
rp = RobotFileParser()
rp.parse(resp_robots.text.splitlines())

boleh_scrape = rp.can_fetch(user_agent, url_scrape)
print("=== VERIFIKASI ROBOTS.TXT DENGAN CAN_FETCH (SOAL 4) ===")
print(f"URL robots.txt : {url_robots} (Status: {resp_robots.status_code})")
print(f"URL Target     : {url_scrape}")
print(f"User-Agent     : {user_agent}")
print(f"Hasil can_fetch: {boleh_scrape}\n")

assert boleh_scrape, "Scraping dilarang oleh robots.txt!"

time.sleep(1.5)
resp_page = requests.get(url_scrape, headers=headers, timeout=15)

if resp_page.status_code == 200:
    raw_html_path = "data raw/scraping_provinsi_indonesia_raw.html"
    with open(raw_html_path, "w", encoding="utf-8") as f:
        f.write(resp_page.text)

    with open(raw_html_path, "rb") as f:
        md5_sumber3 = hashlib.md5(f.read()).hexdigest()

    print("=== PROVENANCE SUMBER 3 (WEB SCRAPING) ===")
    print(f"Berkas Mentah : {raw_html_path}")
    print(f"Tanggal Akses : 2026-11-27")
    print(f"Lisensi       : CC BY-SA 4.0 (Wikipedia)")
    print(f"MD5 Hash      : {md5_sumber3}\n")

    soup = BeautifulSoup(resp_page.text, "html.parser")
    tabel_prov = soup.find("table", {"class": "wikitable"})
    df_prov_raw = pd.read_html(StringIO(str(tabel_prov)))[0]

    df_prov_bersih = df_prov_raw.copy()
    df_prov_bersih.columns = [str(c).strip() for c in df_prov_bersih.columns]
    kolom_nama_prov = [c for c in df_prov_bersih.columns if "Provinsi" in c][0]
    df_prov_bersih["kunci_provinsi"] = (
        df_prov_bersih[kolom_nama_prov]
        .astype(str)
        .str.replace(r"\[.*?\]", "", regex=True)
        .str.strip()
        .str.title()
    )
    df_prov_bersih = df_prov_bersih.drop_duplicates(subset=["kunci_provinsi"])

    df_prov_bersih.to_csv("process/scraping_provinsi_indonesia_bersih.csv", index=False, encoding="utf-8")
    print(f"Berhasil mengekstrak {len(df_prov_bersih)} provinsi di Indonesia!")
    display(df_prov_bersih.head(10))
else:
    print(f"Gagal mengakses halaman target. Status: {resp_page.status_code}")
```

## Soal 5: Penggabungan Data (`pd.merge`) & Validasi Anti-Duplikasi

### 1. Strategi & Kunci Penggabungan Berjenjang
Untuk mengatasi keterbatasan cakupan wilayah pada versi sebelumnya (yang hanya memuat data administratif Provinsi Bali), penggabungan data pada tahap ini dilakukan dalam dua tahapan secara berjenjang:
1. **Tahap 1 — Wisata Nusantara (`df_wisata_bersih`) × Cuaca & Elevasi API (`df_api_bersih`):**
   * Menggunakan kunci penggabungan **`kunci_wilayah`** (nama kabupaten/kota yang telah dinormalisasi spasi dan kapitalisasinya) dengan metode `how="inner"` dan validasi relasi **`validate="many_to_one"`**.
2. **Tahap 2 — Hasil Gabungan Tahap 1 × Administrasi Provinsi Indonesia (`df_prov_rapi`):**
   * Menggunakan kunci penggabungan **`kunci_provinsi`** dengan metode `how="left"` dan validasi relasi **`validate="many_to_one"`**, sehingga seluruh destinasi wisata dari berbagai provinsi di Indonesia mendapatkan atribut administratif provinsi (ibu kota, wilayah geografis, dan jumlah kabupaten/kota) secara merata.

### 2. Pemeriksaan Integritas & Anti-Duplikasi
* Memastikan tabel referensi API (`kunci_wilayah`) dan hasil *scraping* provinsi (`kunci_provinsi`) bersifat unik (1 baris per wilayah) sebelum digabungkan.
* Melakukan verifikasi jumlah duplikasi baris setelah proses *merge* menggunakan fungsi `df.duplicated().sum()` serta menyimpan hasil akhir ke `process/data_gabungan_wisata_final.csv` dan berkas metadatanya ke `process/metadata_data_gabungan.json`.
*

```python
import json
import pandas as pd

df_prov_rapi = df_prov_bersih.copy()
df_prov_rapi.columns = [
    c[0] if isinstance(c, tuple) else str(c).strip()
    for c in df_prov_rapi.columns
]

if "Lambang" in df_prov_rapi.columns:
    df_prov_rapi = df_prov_rapi.drop(columns=["Lambang"])

for col in df_prov_rapi.columns:
    if df_prov_rapi[col].dtype == object:
        df_prov_rapi[col] = (
            df_prov_rapi[col]
            .astype(str)
            .str.replace(r"\[.*?\]", "", regex=True)
            .str.strip()
        )

df_prov_rapi = df_prov_rapi.drop_duplicates(subset=["kunci_provinsi"])
df_prov_rapi.to_csv("process/scraping_provinsi_indonesia_bersih.csv", index=False, encoding="utf-8")

df_s1 = df_wisata_bersih.copy()
df_s1["kunci_wilayah"] = (
    df_s1["kota_kabupaten"]
    .astype(str)
    .str.replace(r"^(Kabupaten|Kota)\s+", "", regex=True)
    .str.strip()
    .str.title()
)
df_s1["kunci_provinsi"] = df_s1["provinsi"].astype(str).str.strip().str.title()

df_m1 = pd.merge(
    df_s1,
    df_api_bersih.drop_duplicates(subset=["kunci_wilayah"]),
    on="kunci_wilayah",
    how="inner",
    validate="many_to_one"
)

kolom_prov_ambil = ["kunci_provinsi"] + [
    c for c in df_prov_rapi.columns
    if any(k in c for k in ["Ibu kota", "Wilayah geografis", "Luas", "Populasi", "Penduduk", "Tingkat II"])
]
df_gabungan_final = pd.merge(
    df_m1,
    df_prov_rapi[kolom_prov_ambil],
    on="kunci_provinsi",
    how="left",
    validate="many_to_one"
)

jumlah_baris = int(len(df_gabungan_final))
jumlah_kolom = int(len(df_gabungan_final.columns))
jumlah_duplikat = int(df_gabungan_final.duplicated().sum())
assert jumlah_duplikat == 0, "Ditemukan duplikasi baris!"

print("=== VERIFIKASI HASIL MERGE FINAL (SOAL 5) ===")
print(f"Kunci Penggabungan  : kunci_wilayah (Kab/Kota) & kunci_provinsi (Provinsi)")
print(f"Validasi Relasi     : many_to_one (Berhasil tanpa error)")
print(f"Dimensi Hasil Akhir : {jumlah_baris} baris x {jumlah_kolom} kolom")
print(f"Jumlah Duplikasi    : {jumlah_duplikat} baris (Terbukti bebas duplikasi)")
print(f"Cakupan Provinsi    : {df_gabungan_final['kunci_provinsi'].nunique()} provinsi terisi merata\n")

df_gabungan_final.to_csv("process/data_gabungan_wisata_final.csv", index=False, encoding="utf-8")

metadata_final = {
    "judul": "Dataset Terintegrasi Pariwisata Nusantara, Meteorologi-Elevasi API, dan Provinsi Indonesia",
    "mahasiswa": "Yustina Dhambo (2501010337 - Kelas F)",
    "tanggal_pembuatan": "2026-11-27",
    "sumber_data": [
        "Kaggle - Indonesia Tourism (wisata_nusantara.csv)",
        "Open-Meteo Public API (api_cuaca_elevasi_raw.json)",
        "Wikipedia Indonesia - Provinsi di Indonesia (scraping_provinsi_indonesia_raw.html)"
    ],
    "kunci_penggabungan": ["kunci_wilayah", "kunci_provinsi"],
    "validasi_merge": "many_to_one",
    "jumlah_baris": jumlah_baris,
    "jumlah_kolom": jumlah_kolom,
    "jumlah_duplikasi": jumlah_duplikat
}
with open("process/metadata_data_gabungan.json", "w", encoding="utf-8") as f:
    json.dump(metadata_final, f, indent=4, ensure_ascii=False)

display(df_gabungan_final.head(10))
```
